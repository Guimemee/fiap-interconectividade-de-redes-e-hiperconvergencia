### Formulação Matemática Formal
$$\text{RTT} = t_{\text{ACK}} - t_{\text{SYN}}, \quad \text{Taxa de Retransmissão} = \frac{N_{\text{Retrans}}}{N_{\text{Total}}} \times 100\%$$


### Passo 1: Aplicação de Filtros de Captura e Exibição
Filtro: tcp.analysis.retransmission || tcp.analysis.duplicate_ack para isolamento de problemas.


### Passo 2: Análise do Three-Way Handshake TCP
Medição do RTT inicial (SYN -> SYN/ACK -> ACK) de 14 ms.


### Passo 3: Detecção de Gargalo de Janela TCP (Zero Window)
Identificação de receptores com buffer esgotado anunciando Window Size = 0.


### Passo 4: Análise Estatística de Conversações e Gráficos IO
Gráfico de Throughput revelando perda de pacotes exatamente nos picos de backup noturno.


### Passo 5: Conclusão Diagnóstica
Resolução do MTU mismatch de 1500 para 1492 eliminando fragmentação excessiva de pacotes.


```python
import numpy as np
import matplotlib.pyplot as plt

tempo_s = np.arange(0, 10)
trafego_normal = np.array([12, 14, 15, 18, 16, 17, 19, 15, 14, 16])
retransmissoes = np.array([0, 1, 0, 8, 14, 11, 2, 0, 1, 0])
fig, ax = plt.subplots(figsize=(6, 3.5), facecolor='#14102c')
ax.set_facecolor('#14102c')
ax.plot(tempo_s, trafego_normal, color='#6366f1', lw=2, label='Vazão Útil (MB/s)')
ax.plot(tempo_s, retransmissoes, color='#ef4444', lw=2.5, label='Retransmissões TCP (Spike de Perda)')
ax.set_title('Análise de Tráfego Wireshark no Incidente de Rede', color='#ffffff', fontsize=11, fontweight='bold')
ax.set_xlabel('Tempo da Captura (segundos)', color='#94a3b8')
ax.set_ylabel('Pacotes / Megabytes por Segundo', color='#94a3b8')
ax.tick_params(colors='#94a3b8')
ax.grid(True, color='#352a6b', alpha=0.5)
ax.legend(facecolor='#211a45', edgecolor='#6366f1', labelcolor='#ffffff')
plt.tight_layout()
plt.savefig('grafico_resolucao.png', dpi=120)
plt.close()
```
