> **Enunciado Original:**
> Dimensionar um cluster HCI de 4 nós com armazenamento definido por software (SDS), calculando a capacidade útil sob políticas RAID-1 (Espelhamento) e RAID-5 (Codificação de Apagamento).

### Formulação Matemática Formal
$$\text{Capacidade Útil (RAID-1)} = \frac{\text{Bruto}}{2}, \quad \text{Capacidade Útil (RAID-5)} = \text{Bruto} \times \left(\frac{N - 1}{N}\right) = \text{Bruto} \times 0,75$$


### Passo 1: Agregação de Recursos de Computação e Armazenamento
Cada nó provê 64 vCPUs, 256 GB RAM e 8 TB de armazenamento NVMe flash.


### Passo 2: Definição do Fator de Falhas Toleradas (FTT)
Política FTT=1 permite a perda catastrófica de 1 nó ou disco inteiro sem indisponibilidade.


### Passo 3: Comparativo de Políticas de Armazenamento
Para 32 TB brutos: RAID-1 provê 16 TB úteis; RAID-5 erasure coding provê 24 TB úteis.


### Passo 4: Configuração de Rede Dedicada de Armazenamento
Dois enlaces 25 GbE redundantes com suporte a Jumbo Frames (MTU 9000).


### Passo 5: Conclusão de Alta Disponibilidade
Cluster com failover automático vSphere HA com tempo de recuperação de VMs inferior a 20 segundos.


```python
import numpy as np
import matplotlib.pyplot as plt

politicas = ['Bruto Total', 'RAID-1 (Espelho)', 'RAID-5 (Erasure Coding)']
capacidade = [32, 16, 24]
cores = ['#94a3b8', '#6366f1', '#10b981']
fig, ax = plt.subplots(figsize=(6, 3.5), facecolor='#14102c')
ax.set_facecolor('#14102c')
bars = ax.bar(politicas, capacidade, color=cores, width=0.45)
ax.set_title('Capacidade de Armazenamento Útil do Cluster HCI (TB)', color='#ffffff', fontsize=11, fontweight='bold')
ax.set_ylabel('Capacidade de Dados (Terabytes)', color='#94a3b8')
ax.tick_params(colors='#94a3b8')
ax.grid(axis='y', color='#352a6b', alpha=0.5)
for bar in bars:
    yval = bar.get_height()
    ax.text(bar.get_x() + bar.get_width()/2.0, yval + 0.8, f'{yval} TB', ha='center', va='bottom', color='#ffffff', fontweight='bold')
plt.tight_layout()
plt.savefig('grafico_resolucao.png', dpi=120)
plt.close()
```
