### Formulação Matemática Formal
$$\text{Custo OSPF} = \frac{\text{Auto-Cost Reference Bandwidth}}{\text{Interface Bandwidth}} = \frac{100.000.000 \text{ bps}}{\text{Bandwidth}}$$


### Passo 1: Estabelecimento de Adjacências de Vizinhos
Troca de pacotes Hello a cada 10 segundos para eleição de DR (Designated Router) e BDR.


### Passo 2: Inundação de LSAs (Link-State Advertisements)
Propagação de LSA Tipo 1 (Router) e Tipo 2 (Network) na base LSDB.


### Passo 3: Execução do Algoritmo SPF (Shortest Path First)
Dijkstra calculando a árvore de menor custo para cada destino na tabela de roteamento.


### Passo 4: Sintonização da Largura de Banda de Referência
auto-cost reference-bandwidth 10000 para diferenciar corretamente enlaces Gigabit e 10-Gigabit.


### Passo 5: Conclusão de Convergência
Tempo de convergência da malha inferior a 2,3 segundos em caso de falha de fibra óptica.


```python
import numpy as np
import matplotlib.pyplot as plt

cenarios = ['Sem Ajuste (Padrão)', 'Com Sintonização BFD/Timers']
convergencia_s = [38.5, 1.8]
fig, ax = plt.subplots(figsize=(6, 3.5), facecolor='#14102c')
ax.set_facecolor('#14102c')
bars = ax.bar(cenarios, convergencia_s, color=['#ef4444', '#10b981'], width=0.4)
ax.set_title('Tempo de Convergência de Falha OSPF (segundos)', color='#ffffff', fontsize=11, fontweight='bold')
ax.set_ylabel('Tempo de Convergência (s)', color='#94a3b8')
ax.tick_params(colors='#94a3b8')
ax.grid(axis='y', color='#352a6b', alpha=0.5)
for bar in bars:
    yval = bar.get_height()
    ax.text(bar.get_x() + bar.get_width()/2.0, yval + 0.8, f'{yval}s', ha='center', va='bottom', color='#ffffff', fontweight='bold')
plt.tight_layout()
plt.savefig('grafico_resolucao.png', dpi=120)
plt.close()
```
