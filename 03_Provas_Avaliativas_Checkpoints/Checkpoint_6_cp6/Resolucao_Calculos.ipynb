### Formulação Matemática Formal
$$\text{Throughput IPsec} = \frac{\text{Payload MTU} - \text{ESP Headers}}{\text{Payload MTU}} \times \text{Bandwidth}, \quad \text{AES-256-GCM}$$


### Passo 1: Negociação IKEv2 Fase 1
Autenticação mútua via certificados digitais X.509 e troca Diffie-Hellman Grupo 19 (Curva Elíptica).


### Passo 2: Negociação IKEv2 Fase 2 (IPsec SA)
Estabelecimento do encapsulamento ESP com cifras AES-256-GCM para criptografia e integridade simultâneas.


### Passo 3: Política de Roteamento Baseada em Aplicação (SD-WAN)
Tráfego crítico de telemetria direcionado via link satelital de baixa latência; backups via link móvel 5G.


### Passo 4: Monitoramento Contínuo de SLA de Enlaces
Sondas BFD medindo jitter, perda de pacotes e latência em tempo real a cada 500 ms.


### Passo 5: Conclusão de Segurança e Desafio Sanofi
Comunicação de dados marítimos impenetrável com comutação transparente entre operadoras sem queda de sessão.


```python
import numpy as np
import matplotlib.pyplot as plt

links = ['Link Satelital Primário', 'Link 5G Redundante', 'Link Backup HF']
latencias = [22.5, 48.0, 110.0]
fig, ax = plt.subplots(figsize=(6, 3.5), facecolor='#14102c')
ax.set_facecolor('#14102c')
bars = ax.bar(links, latencias, color='#a855f7', width=0.45)
ax.set_title('Latência Fim-a-Fim por Enlace SD-WAN (ms)', color='#ffffff', fontsize=11, fontweight='bold')
ax.set_ylabel('Latência Média RTT (ms)', color='#94a3b8')
ax.tick_params(colors='#94a3b8')
ax.grid(axis='y', color='#352a6b', alpha=0.5)
for bar in bars:
    yval = bar.get_height()
    ax.text(bar.get_x() + bar.get_width()/2.0, yval + 2, f'{yval} ms', ha='center', va='bottom', color='#ffffff', fontweight='bold')
plt.tight_layout()
plt.savefig('grafico_resolucao.png', dpi=120)
plt.close()
```
