### Formulação Matemática Formal
$$\text{Hosts Necessários} \le 2^{32 - \text{prefixo}} - 2, \quad \text{Ex: } /27 \implies 2^5 - 2 = 30 \text{ hosts}$$


### Passo 1: Inventário de Requisitos por Segmento
Setor Farmacêutico: 120 hosts (/25); Sensores IoT: 60 hosts (/26); Servidores: 28 hosts (/27); Enlaces ponto-a-ponto: 2 hosts (/30).


### Passo 2: Alocação Hierárquica a partir do Bloco 10.20.0.0/22
Ordenação decrescente por quantidade de hosts requeridos.


### Passo 3: Cálculo de Máscaras e Broadcast
Sub-rede 1 (Produção): 10.20.0.0/25 (Faixa: 10.20.0.1 a 10.20.0.126; Broadcast: 10.20.0.127).


### Passo 4: Planejamento Duplo com IPv6 Global Unicast (/64)
Atribuição de prefixos /64 com autoconfiguração SLAAC.


### Passo 5: Conclusão de Eficiência
Aproveitamento de 92% do espaço de endereçamento com espaço garantido para expansão futura.


```python
import numpy as np
import matplotlib.pyplot as plt

subredes = ['Produção (/25)', 'IoT Sensores (/26)', 'Servidores (/27)', 'Enlaces (/30)']
ips_alocados = [128, 64, 32, 4]
cores = ['#6366f1', '#a855f7', '#818cf8', '#38bdf8']
fig, ax = plt.subplots(figsize=(6, 3.5), facecolor='#14102c')
ax.set_facecolor('#14102c')
bars = ax.bar(subredes, ips_alocados, color=cores, width=0.5)
ax.set_title('Alocação de Endereços por Sub-rede VLSM', color='#ffffff', fontsize=11, fontweight='bold')
ax.set_ylabel('Espaço de IPs Alocado', color='#94a3b8')
ax.tick_params(colors='#94a3b8')
ax.grid(axis='y', color='#352a6b', alpha=0.5)
for bar in bars:
    yval = bar.get_height()
    ax.text(bar.get_x() + bar.get_width()/2.0, yval + 2, str(yval), ha='center', va='bottom', color='#ffffff', fontweight='bold')
plt.tight_layout()
plt.savefig('grafico_resolucao.png', dpi=120)
plt.close()
```
