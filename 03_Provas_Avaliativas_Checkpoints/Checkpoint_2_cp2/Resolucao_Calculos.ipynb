> **Enunciado Original:**

### Formulação Matemática Formal
$$\text{Header 802.1Q}: \text{TPID (16 bits: 0x8100)} + \text{TCI (Priority: 3 bits, CFI: 1 bit, VID: 12 bits)}$$


### Passo 1: Criação das VLANs no Switch Core
VLAN 10 (Gestao), VLAN 20 (Sensores), VLAN 30 (DMZ), VLAN 99 (Nativa).


### Passo 2: Configuração dos Enlaces Trunk 802.1Q
switchport mode trunk; switchport trunk allowed vlan 10,20,30,99.


### Passo 3: Subinterfaces de Roteamento no Roteador
interface GigabitEthernet0/0/0.10; encapsulation dot1Q 10; ip address 10.20.0.1 255.255.255.128.


### Passo 4: Regras de Segurança com ACL Estendida
Bloqueio de acesso da VLAN de Sensores aos servidores da VLAN de Gestão via porta 80/443.


### Passo 5: Conclusão de Isolamento
Domínio de broadcast reduzido em 75% e contenção total de tempestades de broadcast.


```python
import numpy as np
import matplotlib.pyplot as plt

vlans = ['VLAN 10 (Gestão)', 'VLAN 20 (Sensores)', 'VLAN 30 (DMZ)', 'VLAN 99 (Nativa)']
trafego_mbps = [45, 12, 85, 2]
fig, ax = plt.subplots(figsize=(6, 3.5), facecolor='#14102c')
ax.set_facecolor('#14102c')
bars = ax.barh(vlans, trafego_mbps, color='#6366f1', height=0.45)
ax.set_title('Vazão de Tráfego Segmentado por VLAN (Mbps)', color='#ffffff', fontsize=11, fontweight='bold')
ax.set_xlabel('Tráfego Médio (Mbps)', color='#94a3b8')
ax.tick_params(colors='#94a3b8')
ax.grid(axis='x', color='#352a6b', alpha=0.5)
for bar in bars:
    wval = bar.get_width()
    ax.text(wval + 1, bar.get_y() + bar.get_height()/2.0, f'{wval} Mbps', ha='left', va='center', color='#ffffff', fontweight='bold')
plt.tight_layout()
plt.savefig('grafico_resolucao.png', dpi=120)
plt.close()
```
